# Warp energy-density explorer

Sliders over the Eulerian energy density of a warp bubble, computed from the
Hamiltonian and momentum constraints on flat spatial slices.

Every number here is the same quantity `particlesim run` writes to `report.json`;
the physics lives in `particlesim.viz.widgets.explorer_state`, which this notebook
only wraps. Nothing in the Alcubierre family has positive energy density anywhere,
which is the point the map is making.


```python
from particlesim.viz.widgets import warp_explorer

warp_explorer()
```

## Without widgets

`explorer_state` is a plain function, so the same numbers are available in a
script, a test, or a parameter sweep with no notebook involved.


```python
from particlesim.viz.widgets import explorer_state

for v in (0.5, 1.0, 2.0, 4.0):
    s = explorer_state("alcubierre", v_s=v, n=40).summary()
    print(f"v = {v:>4}: negative energy {s['negative_energy']:>12.4g}")
```
